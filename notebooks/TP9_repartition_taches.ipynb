## Étape 1 — Contrat commun
**13h30 → 13h50**

### Sacha
- Créer le squelette de `src/cleaning.py`.
- Préparer le contrat commun des fonctions de nettoyage.
- Préparer les fixtures/tests de base.

### Clément
- Relire et valider le contrat avant le développement parallèle.
- Préparer une fixture contenant les cas tordus identifiés au TP2, notamment : valeurs nutritionnelles aberrantes, doublon de code-barres, code avec espaces, nom mal normalisé, catégorie vide et produit avec données nutritionnelles absentes.

### Résultat attendu
- Même interface de nettoyage pour les deux branches.
- Cas de test communs identifiés avant l'implémentation.


| Cas                                  | Constat                                                                                                                       | Contrat                                                                                               | Fixture                                              |
| ------------------------------------ | ----------------------------------------------------------------------------------------------------------------------------- | ----------------------------------------------------------------------------------------------------- | ---------------------------------------------------- |
| **Valeur nutritionnelle aberrante**  | Certaines valeurs sont impossibles ou incohérentes, par exemple des sucres > 100 g/100 g ou des valeurs négatives.            | Corriger les valeurs lorsqu'une correction est possible, sinon les mettre à `NULL`.                   | Produit avec une valeur nutritionnelle aberrante.    |
| **Doublon de code-barres**           | Certains produits ont le même code-barres.                                                                                    | Garder une seule version du produit selon les règles définies.                                        | Deux produits avec le même code-barres.              |
| **Code avec espaces**                | Certains codes-barres contiennent des espaces inutiles.                                                                       | Supprimer les espaces pour normaliser le code.                                                        | Produit avec un code contenant des espaces.          |
| **Nom mal normalisé**                | Certains noms sont écrits différemment alors qu'ils désignent le même nom, par exemple `Taste of Nature` / `Taste Of Nature`. | Normaliser le nom selon une règle commune.                                                            | Produit avec un nom mal formaté.                     |
| **Catégorie vide**                   | Beaucoup de produits n'ont pas de catégorie renseignée.                                                                       | Conserver le produit et gérer l'absence de catégorie.                                                 | Produit sans catégorie.                              |
| **Données nutritionnelles absentes** | Certains produits n'ont pas toutes leurs données nutritionnelles.                                                             | Conserver le produit, mais ne pas l'utiliser pour les traitements nécessitant les données manquantes. | Produit avec des données nutritionnelles manquantes. |
