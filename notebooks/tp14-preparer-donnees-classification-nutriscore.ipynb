# TP14 — Préparer les jeux de classification du Nutri-Score

**NutriScope · 20 octobre 2026**

## Objectif et contexte

Le TP14 lance la première expérimentation de classification A–E. L'export ML comporte déjà une colonne `split` figée. Avant d'entraîner des modèles, il faut construire des features sans fuite de cible et garder à part les informations nécessaires à l'analyse des erreurs. Cette étape ne réentraîne pas le split et ne contacte pas PostgreSQL.

## Comportement attendu et critères d'acceptation

- Séparer les observations selon `split` (`train` ou `test`) sans modifier leur affectation.
- Utiliser `nutriscore_grade` comme cible, avec les classes `a`, `b`, `c`, `d`, `e`.
- Retenir les features du cadrage TP14 : sept nutriments obligatoires, `fiber_100g`, `fruits-vegetables-legumes_100g`, `nova_group`, `additives_n`, `pnns_groups_1`, `pnns_groups_2`.
- Exclure des features `nutriscore_grade`, `nutriscore_score`, `code`, `product_name`, `brands` et `split`. Garder les métadonnées utiles séparément pour l'analyse.
- Conserver les valeurs manquantes des features optionnelles pour que l'imputation soit ensuite apprise dans un pipeline sklearn ajusté sur le train seulement.
- Signaler clairement les colonnes obligatoires absentes, les grades ou splits inconnus et une partition train/test vide.

## Interface à écrire

Dans `src/ml_nutriscore.py`, écrire `preparer_jeux_classification(df)`. Son rôle est de valider le contrat de données et de retourner un dictionnaire contenant `X_train`, `y_train`, `X_test`, `y_test` et `donnees_test` (métadonnées destinées à l'analyse des erreurs). Le choix des étapes internes vous appartient ; ne faites ni entraînement ni imputation dans cette fonction.

## Progression

### Étape 1 — Décrire le contrat d'entrée et les sorties

Inspectez `docs/tp-14-ML_README.md`, `docs/tp-14-predire-le-nutriscore.md`, `src/` et les tests. Vérifiez le chemin réel de l'export dans ce checkout et les noms/valeurs des colonnes avant de coder. Écrivez uniquement la signature de fonction et sa docstring en français.

**Point de contrôle :** expliquez quelles colonnes sont features, cible ou métadonnées, et pourquoi `nutriscore_score` ne doit pas entrer dans `X`. Faites valider ce raisonnement avant la suite.

### Étape 2 — Séparer les partitions figées

Construisez les quatre sorties train/test en conservant les index d'origine. Gardez les champs nécessaires au diagnostic séparés des features. Exécutez les tests d'acceptation ciblés et analysez toute divergence.

**Point de contrôle :** démontrez que le split fourni n'a pas été recalculé et que chaque ligne reste dans une seule partition. Faites valider votre code avant la suite.

### Étape 3 — Valider les limites et les erreurs

Traitez explicitement les colonnes absentes, les classes et splits inattendus, ainsi que les partitions vides. Gardez les manquants des features optionnelles : l'imputation viendra avec le pipeline de modèle. Relancez les tests ciblés.

**Point de contrôle :** expliquez pourquoi une entrée invalide échoue explicitement plutôt que d'être corrigée ou écartée silencieusement. Faites valider avant de passer à la construction du pipeline sklearn.
